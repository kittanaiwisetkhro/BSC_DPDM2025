```python
def read_all_2025_rainfall_data(input_folder):
    all_files = []
    # Iterate through months 07 to 12
    for month_num in range(7, 13):
        month_str = f'2025-{month_num:02d}'
        month_folder = os.path.join(input_folder, month_str)

        # Check if the month folder exists
        if os.path.exists(month_folder):
            # Find all CSV files in the current month's folder
            csv_files = glob.glob(os.path.join(month_folder, '*.csv'))
            all_files.extend(csv_files)
        else:
            print(f"Warning: Folder {month_folder} not found. Skipping.")

    df_list = []
    for filename in all_files:
        try:
            df = pd.read_csv(filename)
            df_list.append(df)
        except Exception as e:
            print(f"Error reading {filename}: {e}")

    if df_list:
        combined_df = pd.concat(df_list, ignore_index=True)
        return combined_df
    else:
        print("No files found or read.")
        return pd.DataFrame()

# Call the function to read all 2025 rainfall data
df_2025 = read_all_2025_rainfall_data(input_2025_folder)


df_2025.info()
```

### คำถามที่ 3. ระบุข้อมูลสูญหาย (record ที่หายไป (ไม่ใช่ค่า NULL หรือ NaN)) ของแต่ละสถานี แสดงจำนวนและสัดส่วน

```python
# --- 3.1 เตรียมข้อมูลวันที่ให้เป็น format datetime ---
# แปลง column measure_datetime เป็น datetime object และตัดเอาแค่วันที่ (ตัดเวลาทิ้ง) เพื่อการเปรียบเทียบ
df_2025_filtered_stations['measure_date'] = pd.to_datetime(df_2025_filtered_stations['measure_datetime']).dt.date

# กำหนดช่วงเวลาที่ควรจะมีข้อมูล (ตาม Folder ที่ระบุคือ 2025-07 ถึง 2025-12)
start_date = pd.to_datetime('2025-07-01').date()
end_date = pd.to_datetime('2025-12-31').date()
full_date_range = pd.date_range(start=start_date, end=end_date, freq='D').date

# จำนวนวันที่ควรจะมีทั้งหมด
total_expected_days = len(full_date_range)
print(f"Total expected days (Jul-Dec 2025): {total_expected_days} days")

# --- 3.2 สร้างตารางหลัก (Master Template) ที่มีครบทุกสถานีและทุกวัน ---
unique_stations = df_2025_filtered_stations['tele_station_oldcode'].unique()

# สร้าง MultiIndex ที่เป็นการจับคู่กันระหว่าง ทุกสถานี x ทุกวัน (Cartesian Product)
index = pd.MultiIndex.from_product([unique_stations, full_date_range], names=['tele_station_oldcode', 'measure_date'])
df_full_template = pd.DataFrame(index=index).reset_index()

# --- 3.3 นำข้อมูลจริงมา Merge กับตารางหลัก ---
# ทำ Left Join: ยึดตารางหลักไว้ ถ้าวันไหนในข้อมูลจริงไม่มี ค่า rainfall_value จะกลายเป็น NaN
# ก่อน Merge ต้องจัดการ Duplicate ในข้อมูลจริงก่อน (เผื่อวันหนึ่งมีหลาย record) โดยเลือกค่าใดค่าหนึ่ง หรือค่าเฉลี่ย (ในที่นี้ขอเลือก record แรกที่เจอเพื่อเช็คการมีอยู่)
df_actual_unique = df_2025_filtered_stations.drop_duplicates(subset=['tele_station_oldcode', 'measure_date'])

df_merged = pd.merge(df_full_template,
                     df_actual_unique[['tele_station_oldcode', 'measure_date', 'rainfall_value']],
                     on=['tele_station_oldcode', 'measure_date'],
                     how='left')

# --- 3.4 คำนวณจำนวนและสัดส่วนข้อมูลที่หายไป ---
# สร้างตารางสรุปผล
missing_stats = df_merged.groupby('tele_station_oldcode').apply(
    lambda x: pd.Series({
        'total_expected': len(x),
        'missing_count': x['rainfall_value'].isna().sum(), # นับจำนวนวันที่ไม่มีข้อมูล (NaN หลังจากการ Merge)
        'missing_percent': (x['rainfall_value'].isna().sum() / len(x)) * 100
    })
).reset_index()

# แสดงตัวอย่างผลลัพธ์
print("\n--- Missing Data Statistics per Station ---")
print(missing_stats.head(10))

# แสดงสถิติภาพรวม
print("\n--- Summary Statistics of Missing Data ---")
print(missing_stats[['missing_count', 'missing_percent']].describe())
```

## ขั้นตอนที่ 2. ลบสถานีที่มีข้อมูลสูญหาย มากกว่าหรือเท่ากับ 20%

### คำถามที่ 4. คัดสถานีที่มีข้อมูลสูญหาย มากกว่าหรือเท่ากับ 20% ออก แสดงรายชื่อสถานีที่ถูกคัดออกและจำนวนที่เหลือ

```python
# --- 4.1 กำหนดเกณฑ์และแยกกลุ่มสถานี ---
THRESHOLD_PERCENT = 20.0

# หา List สถานีที่ต้อง "คัดออก" (Missing >= 20%)
stations_to_drop = missing_stats[missing_stats['missing_percent'] >= THRESHOLD_PERCENT]
dropped_list = stations_to_drop['tele_station_oldcode'].tolist()

# หา List สถานีที่ "เก็บไว้" (Missing < 20%)
stations_to_keep = missing_stats[missing_stats['missing_percent'] < THRESHOLD_PERCENT]
kept_list = stations_to_keep['tele_station_oldcode'].tolist()

# --- 4.2 กรอง DataFrame หลักให้เหลือเฉพาะสถานีที่เก็บไว้ ---
# เราใช้ df_merged จากขั้นตอนที่ 3 เพราะตารางนี้มี row ของวันที่หายไปเตรียมไว้ให้ (เป็น NaN) พร้อมสำหรับขั้นตอนที่ 5
df_clean_for_impute = df_merged[df_merged['tele_station_oldcode'].isin(kept_list)].copy()

# --- 4.3 แสดงผลลัพธ์ ---
print(f"Criteria: Remove stations with missing data >= {THRESHOLD_PERCENT}%")
print("-" * 60)
print(f"Number of stations DROPPED : {len(dropped_list)}")
print(f"Number of stations REMAINING: {len(kept_list)}")
print("-" * 60)

print("List of dropped stations (Code):")
if len(dropped_list) > 0:
    print(dropped_list)
else:
    print("None (No stations exceeded the missing threshold)")

print("\nShape of data ready for imputation:", df_clean_for_impute.shape)
```

## ขั้นตอนที่ 3. เติมข้อมูลด้วย Median ของ Day of Year จาก historical

คำถามที่ 5. เติมข้อมูลที่หายไปด้วยค่า Median ตาม Day of Year จาก historical แยกตามสถานี ตรวจสอบว่าไม่มี record หายอีก


```python
# --- 5.1 เตรียมข้อมูล Historical (หา Median ตาม Day of Year) ---
# แปลงเป็น datetime (ถ้ายังไม่เป็น)
df_history['measure_datetime'] = pd.to_datetime(df_history['measure_datetime'])

# สร้าง column Day of Year (1-366)
df_history['doy'] = df_history['measure_datetime'].dt.dayofyear

# คำนวณ Median แยกตาม สถานี (station_code) และ วันของปี (doy)
# ใช้ 'data' column ซึ่งคือปริมาณฝนใน df_history
historical_medians = df_history.groupby(['station_code', 'doy'])['data'].median().reset_index()
historical_medians.rename(columns={'data': 'median_val'}, inplace=True)

# --- 5.2 เตรียมข้อมูล 2025 ที่จะเติม (df_clean_for_impute) ---
# แปลง measure_date กลับเป็น datetime เพื่อหา doy
df_clean_for_impute['measure_date'] = pd.to_datetime(df_clean_for_impute['measure_date'])
df_clean_for_impute['doy'] = df_clean_for_impute['measure_date'].dt.dayofyear

# --- 5.3 เติมข้อมูล (Imputation) ---
# Merge ค่า Median เข้ามาในตารางหลัก โดยใช้ station code และ doy เป็น key
# หมายเหตุ: ใน df_history ใช้ 'station_code', ใน df_2025 ใช้ 'tele_station_oldcode'
df_imputed = pd.merge(df_clean_for_impute,
                      historical_medians,
                      left_on=['tele_station_oldcode', 'doy'],
                      right_on=['station_code', 'doy'],
                      how='left')

# สร้าง column ใหม่ 'rainfall_filled'
# ถ้ามีค่าเดิมอยู่แล้วให้ใช้ค่าเดิม, ถ้าเป็น NaN ให้ใช้ค่า median_val ที่ดึงมา
df_imputed['rainfall_filled'] = df_imputed['rainfall_value'].fillna(df_imputed['median_val'])

# --- 5.4 ตรวจสอบผลลัพธ์ (Verification) ---
# นับจำนวนช่องว่างที่ยังเหลืออยู่ (ในกรณีที่ Historical ก็ไม่มีข้อมูลของวันนั้นๆ เลย)
remaining_nans = df_imputed['rainfall_filled'].isna().sum()
total_records = len(df_imputed)

print(f"--- Imputation Result ---")
print(f"Total records processed: {total_records}")
print(f"Missing values BEFORE imputation: {df_clean_for_impute['rainfall_value'].isna().sum()}")
print(f"Missing values AFTER imputation : {remaining_nans}")

if remaining_nans == 0:
    print("\n✅ SUCCESS: No missing records remaining.")
else:
    print(f"\n⚠️ WARNING: There are still {remaining_nans} missing records.")
    print("This happens if the historical data also lacks data for that specific station and day of year.")
    # (Optional) เติมด้วย 0 หรือค่าเฉลี่ยรวมของสถานีนั้นๆ หากจำเป็น
    # df_imputed['rainfall_filled'].fillna(0, inplace=True)

# แสดงตัวอย่างข้อมูลที่ถูกเติม (เลือกดูเฉพาะแถวที่ค่าเดิมเป็น NaN)
print("\nExample of imputed rows:")
print(df_imputed[df_imputed['rainfall_value'].isna()][['tele_station_oldcode', 'measure_date', 'rainfall_filled', 'median_val']].head())
```

#### 6. อภิปรายว่า เหตุใดใช้ Median แทน Mean

ข้อมูลน้ำฝนมีการกระจายตัวแบบเบ้ขวา (Right-skewed) คือมีวันฝนไม่ตก (0 mm) จำนวนมาก และมีวันฝนตกหนัก (Outliers) จำนวนน้อย

ค่า Mean จะถูกดึงด้วยค่าฝนตกหนัก ทำให้ค่าที่ได้สูงเกินจริงสำหรับวันปกติ

ค่า Median จะระบุค่ากลางที่แท้จริงได้ดีกว่า (Robust to outliers)
