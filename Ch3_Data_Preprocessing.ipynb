```python
# create a function output a mean of bin the value landed in where the bin is following the histogram of 30 bin
```

```python
def get_bin_mean(value, edges):
    """
    รับค่า: value (ตัวเลขที่ต้องการตรวจสอบ), edges (ขอบเขตจาก outputhist[1])
    คืนค่า: จุดกึ่งกลาง (Mean) ของ Bin ที่ค่านั้นตกลงไป
    """
    # 1. ตรวจสอบขอบเขต (Out of bound check)
    if value < edges[0] or value > edges[-1]:
        return "Out of Range (อยู่นอกขอบเขต)"

    # 2. วนลูปหาตำแหน่ง Bin
    for i in range(len(edges) - 1):
        lower = edges[i]
        upper = edges[i+1]

        # เช็คเงื่อนไข: ครอบคลุม Bin สุดท้ายด้วย (<=)
        if (i == len(edges) - 2 and lower <= value <= upper) or \
           (lower <= value < upper):

            # 3. คำนวณค่าเฉลี่ย (จุดกึ่งกลาง) ของ Bin นั้น
            bin_mean = (lower + upper) / 2
            return bin_mean

    return None
```

```python
def put_value_into_bins(value, bin_edge):
  for i in range(len(bin_edge)-1):
      if value >= bin_edge[i] and value < bin_edge[i+1]:
        new_val = bin_edge[i] + (bin_edge[i+1] - bin_edge[i])/2
        return new_val
  return None
```

```python
put_value_into_bins(3000,outputhist[1])
```

```python
df['gfd_area_bin'] = df['gfd_area'].apply(lambda x: put_value_into_bins(x,outputhist[1]))
```
